## Missing Data Handling - Age
The Age column contains 177 missing values.
Median imputation is used because age is a numerical variable and the median is less affected by extreme values than the mean. This approach allows us to retain passenger records rather than deleting them because of missing age information.

```python
age_median = df["Age"].median()
print("median age:", age_median)
df["Age"] = df["Age"].fillna(age_median)
```

```python
df["Age"].isnull().sum()
```

## Missing Data Handling - Cabin
The Cabin column contains 687 missing values, which represents a large proportion of the dataset.
Deleting all rows with missing cabin information would result in substantial information loss. Therefore, missing cabin values are replaced with "unknown". This preserves the passenger records while clearly indicating that cabin information was unavailable.

```python
df["Cabin"] = df["Cabin"].fillna("unknown")
```

```python
df["Cabin"].isnull().sum()
```

## Missing Data Handling - Embarked
The Embarked column contains only 2 missing values and is categorical.
Mode imputation is used because the mode represents the most frequently occuring category. This allows the two affected records to be retained without unnecessarily deleting rows.

```python
embarked_mode = df["Embarked"].mode()[0]
print("Most common Embarked value:", embarked_mode)
df["Embarked"] = df["Embarked"].fillna(embarked_mode)
```

## Standardisation
Text fields are standardised by removing unnecessary leading and trailing spaces and applying consistent formatting.
Sex values are standardised to "Male" and "Female", while Embarked values are standardised to uppercase codes such as S, C, and Q.

```python
# Remove unnecessary spaces
df["Name"] = df["Name"].str.strip()
df["Ticket"] = df["Ticket"].str.strip()
df["Cabin"] = df["Cabin"].str.strip()
# Standardise Sex
df["Sex"] =df["Sex"].str.strip().str.lower()

df["Sex"] = df["Sex"].replace({"m": "male", "f": "female"})
df["Sex"] =df["Sex"].str.title()

# Standardise Embarked
df["Embarked"] = df["Embarked"].str.strip().str.upper()
```

## Outlier Detection Using IQR
The Interquartile Range (IQR) method is used to identify potential outliers in numerical columns. The numerical columns examined are Age, SibSp, Parch, and Fare.
An observation identified as an outlier is not automatically considered an error. The values must be examined before deciding whether to remove, cap, or retain them.

```python
def iqr_outlier_report(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 - 1.5 * IQR

    outliers = data[
    (data[column] < lower_bound) |
    (data[column] > upper_bound)
        ]
    return {
    "column": column,
    "Q1": Q1,
    "Q3": Q3,
    "IQR": IQR,
    "Lower_Bound":lower_bound,
    "Upper_Bound": upper_bound,
    "Outlier_Count": len(outliers)
    }
    numeric_columns = ["Age", "SibSp", "Parch", "Fare"]
    outlier_results = []
    for column in numeric_columns:
        outlier_results.append(iqr_outlier_report(df, column)
                              )
        outlier_report = pd.DataFrame(outlier_results)
        outlier_report
```

## Outlier Handling Decision
The IQR method identified potential outliers in several numerical columns. The identified values were not automatically removed because statiscal outliers are not necessarily data_entry errors. In the Titanic dataset, unussually high fares can represent legitimate ticket purchases, while higher values in SibSp and Parch can represent genuine family relationships.
Age values within the observed range can also represent legitimate passenger ages. Therefore, the identified outliers are retained rather than removed or capped because there is insufficient evidence that they are erroneous values. Retaining them prevents the loss of potentially meaningful observations.

## Data Type Correction
Data types are corrected according to the meaning of each variable. Passengerid, Name, Ticket, and Cabin are treated as string fields.Pclass, Sex, and Embarked are categorical variables. Age, Fare are numerical variables, while Survived, SibSp, and Parch are integer variables.

```python
df["PassengerId"] = df["PassengerId"].astype("string")
df["Survived"] = df["Survived"].astype("int64")
df["Pclass"] = df["Pclass"].astype("category")
df["Name"] = df["Name"].astype("string")
df["Sex"] = df["Sex"].astype("category")
df["Age"] = df["Age"].astype("float64")
df["SibSp"] = df["SibSp"].astype("int64")
df["Parch"] = df["Parch"].astype("int64")
df["Ticket"] = df["Ticket"].astype("string")
df["Fare"] = df["Fare"].astype("float64")
df["Cabin"] = df["Cabin"].astype("string")
df["Embarked"] = df["Embarked"].astype("category")
```

## Dtype Accuracy
The corrected data types are compared with the expected data types to determine the accuracy of the data type correction.

```python
expected_dtypes = {
    "PassengerId": "string",
    "Survived": "int64",
    "Pclass": "category",
    "Name": "string",
    "Sex": "category",
    "Age": "float64",
    "SibSp": "int64",
    "Parch": "int64",
    "Ticket": "string",
    "Fare": "float64",
    "Cabin": "string",
    "Embarked": "category" 
}
dtype_accuracy = sum(
    str(df[column].dtype) == expected_dtype for column, expected_dtype in expected_dtypes.items())

total_columns = len(expected_dtypes)
print("correct data types:", dtype_accuracy,"/", total_columns)
print("Dtype accuracy:",
      round(dtype_accuracy / total_columns * 100, 2),
      "%"
     )
    
    
```

```python
for column, expected_dtype in expected_dtypes.items():
    actual_dtype = str(df[column].dtype)
    if actual_dtype != expected_dtype:
        print(f"{column}: Actual = {actual_dtype}, Expected = {expected_dtype}")
```

```python
after_rows = len(df)
after_nulls =df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()

final_summary = pd.DataFrame({
    "Metric": ["Row count", "Total null count", "Duplicate row count", "Dtype accuracy"],
    "Before Cleaning": [before_rows, before_nulls, before_duplicates, "Required correction"],
    "After Cleaning": [after_rows, after_nulls, after_duplicates, f"{dtype_accuracy}/ {total_columns}( 100%)"]})
final_summary
```
